```python
with open("cars.csv", "r") as file:
    lines = file.readlines()

print(lines[:5])
```

```python
cars = []

for line in lines:
    row = line.strip().split(",")
    cars.append(row)

print(cars[:5])
```

```python
print("Cars with more than 20 MPG:")

for car in cars[1:]:
    mpg = float(car[1])

    if mpg > 20:
        print(car[0], "-", mpg, "MPG")
```
